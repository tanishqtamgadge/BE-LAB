1. Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

```

2. Add Dataset

```python
df = pd.read_csv('City_Air_Quality.csv')

print("Dataset Loaded Successfully!")

print(df.head())
```

4. Data cleaning

```python
# Display first five rows
print(df.head())

# Display dataset information
print(df.info())

# Display dataset shape
print("Dataset Shape:", df.shape)

# Statistical summary
print(df.describe())

# Display column names
print(df.columns.tolist())
```

```python
df = df.drop_duplicates()

df['Date'] = pd.to_datetime(df['Date'])

df = df.dropna()

print("Data Cleaned Successfully!")
```

5. Line Chart – AQI Trend

```python
plt.plot(df['Date'], df['AQI'])

plt.title("AQI Trend")
plt.xlabel("Date")
plt.ylabel("AQI")

plt.xticks(rotation=45)
plt.show()
```

6. Line Chart – PM2.5 and PM10

```python
plt.plot(df['Date'], df['PM2.5'], label='PM2.5')

plt.plot(df['Date'], df['PM10'], label='PM10')

plt.title("Pollutant Levels")
plt.xlabel("Date")
plt.ylabel("Pollutant Level")

plt.legend()
plt.xticks(rotation=45)
plt.show()
```

7. Bar Chart – AQI Values

```python
plt.bar(df['Date'].astype(str), df['AQI'])

plt.title("AQI Bar Chart")
plt.xlabel("Date")
plt.ylabel("AQI")

plt.xticks(rotation=45)
plt.show()
```

8. Histogram – AQI Distribution

```python
plt.hist(df['AQI'], bins=10)

plt.title("AQI Distribution")
plt.xlabel("AQI")
plt.ylabel("Frequency")

plt.show()
```

9. Scatter Plot – AQI vs PM2.5

```python
plt.scatter(df['PM2.5'], df['AQI'])

plt.title("AQI vs PM2.5")
plt.xlabel("PM2.5")
plt.ylabel("AQI")

plt.show()
```

10. Box Plot – AQI

```python
plt.boxplot(df['AQI'])

plt.title("AQI Box Plot")
plt.ylabel("AQI")

plt.show()
```

11. Pie Chart – Pollutant Distribution

```python
values = [
    df['PM2.5'].mean(),
    df['PM10'].mean(),
    df['CO'].mean()
]

labels = ['PM2.5', 'PM10', 'CO']

plt.pie(values, labels=labels, autopct='%1.1f%%')

plt.title("Average Pollutant Distribution")

plt.show()
```

```python
print("Total Records:", len(df))

print("Average AQI:", df['AQI'].mean())

print("Maximum AQI:", df['AQI'].max())

print("Minimum AQI:", df['AQI'].min())

print("Average PM2.5:", df['PM2.5'].mean())

print("Average PM10:", df['PM10'].mean())

print("Air Quality Analysis Completed!")
```
